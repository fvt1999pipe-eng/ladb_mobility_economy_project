<details>

## 🧩Paso 5: Unir movilidad y economía

Combinar datasets te permite analizar cómo se relacionan los indicadores económicos con los de movilidad.

### 5.1 Unir tráfico (tabla principal) con indicadores económicos

**🎯Objetivo:**
Combinar la información de tráfico y economía en un solo DataFrame para analizar cómo las condiciones económicas se relacionan con la movilidad urbana.

**Instrucciones**
- Selecciona solo las **columnas relevantes** de cada dataset (por ejemplo, variables clave de tráfico y de economía).
- Usa `.copy()` al crear subconjuntos para evitar modificar el dataset original.
- Une ambos DataFrames y define como **claves de unión** a `city` y `year`.
- Mantén solo las ciudades y años presentes en ambos datasets.
- Guarda el resultado en una nueva variable llamada `merged` y muestra las primeras 5 filas.


La ciudad con el mayor tiempo promedio de tráfico es Ciudad de México (mexico-city, MEX) con 2833.06 minutos de jams_delay — casi 700 min por encima de Tokyo en segundo lugar.

<details>
<summary>Haz clic para ver la pista</summary>
Aplica una unión de tipo "inner" para mantener las ciudades y años presentes en ambos datasets.

```python
# Seleccionar columnas clave de tráfico y economía
left_cols = ['city','country','year','jams_delay','traffic_index_live',
             'jams_length_kms','jams_count','mins_delay',
             'travel_time_live_per_10kms_mins','travel_time_hist_per_10kms_mins']

right_cols = ['city','year','city_gdp_capita','unemployment_pct','pm25','population']

# Usar .copy() para crear los dos nuevos datasets reducidos
traffic_2024_small = traffic_city_year_2024[left_cols].copy()
eco_2024_small = eco_2024[right_cols].copy()

# Unir datasets
merged = pd.merge(traffic_2024_small, eco_2024_small, on =['city','year'], how='inner')

# Mostrar las primeras 5 filas
# tu código aquí

merged.head(5)

```

```python
# Crear boxplot para observar el comportamiento de los minutos de congestion JamsDelay
# crea tu gráfico
sb.boxplot(data=merged, x='jams_delay', showmeans=True)
# obtener promedio para mostrarlo en título
mean_value = merged['jams_delay'].mean()
plt.title(f'Boxplot de JamsDelay (2024)\nPromedio: {mean_value:.2f}')
plt.show()


```

```python
# Crear histograma para ver la distribución de la economía (city_gdp_capita)

sb.histplot(merged['city_gdp_capita'], kde=True, bins=8, color='steelblue')
plt.title('Distribución del PIB per cápita (2024)')
plt.xlabel('PIB per cápita (USD)')
plt.ylabel('Frecuencia')
plt.show()
```

```python
# Gráfico de barras para comparar jams_delay y city_gdp_capita por ciudad
#merged.plot( ... , y=['jams_delay', 'city_gdp_capita'])
merged.set_index('city')[['city_gdp_capita','jams_delay']].sort_values(by='city_gdp_capita', ascending=False).plot(kind='bar')
plt.title('Distribución de la economia por ciudad')
plt.xlabel('ciudad')
plt.ylabel('PIB per Cápita (USD) vs Retraso en Tráfico (Minutos)')
plt.xticks(rotation=90)
plt.show()
```

**Contexto & objetivo:**

El Banco Interamericano de Desarrollo requiere identificar ciudades prioritarias para inversión en infraestructura de transporte. Esta investigación analiza la relación entre movilidad urbana (congestión medida en minutos de retraso acumulado, tiempos de viaje) y productividad económica (PIB per cápita). La hipótesis central es que ciudades con alta congestión y baja productividad representan oportunidades de inversión con mayor retorno potencial: mejorar la movilidad urbana podría catalizar crecimiento económico y bienestar poblacional.

**Cobertura de datos:**  

El análisis integra datos de 2024 provenientes de dos fuentes: TomTom Traffic Index (métricas de congestión vehicular en tiempo real) y OECD Cities (indicadores económicos). Se analizaron 15 ciudades distribuidas en países de América Latina y el Caribe, incluyendo Montevideo, México City, Buenos Aires, Brasilia, São Paulo, Lima, Río de Janeiro, Curitiba, Porto Alegre, Bogotá, Belo Horizonte, Recife, Salvador, Fortaleza y Santiago. Posterior a limpieza y validación, se alcanzó una cobertura completa sin registros faltantes en variables críticas.

**Metodología (alto nivel):**  

Se importaron y exploraron ambos datasets identificando inconsistencias de formato: fechas como texto, números con separadores de miles y símbolos de porcentaje. Se estandarizaron nombres de columnas a snake_case y se limpiaron valores numéricos eliminando caracteres especiales para asegurar consistencia. Se filtró el año 2024 en ambas fuentes y se agregaron métricas de tráfico por ciudad calculando promedios anuales mediante .groupby().agg(). La integración de movilidad y economía se ejecutó mediante pd.merge() con union INNER sobre ciudad y año. Para validación visual se generaron: (1) boxplot de minutos de congestión (JamsDelay) para detectar distribución y outliers, (2) histograma de PIB per cápita para observar forma de distribución económica, y (3) gráfico comparativo de barras superpuesto para analizar relaciones bivariadas.

**Hallazgos iniciales:**  

No existe una relación lineal clara entre PIB per cápita y congestión vehicular. Se observa una relación inversa en algunos casos: Montevideo (USD 26,500) opera prácticamente sin congestión (~50 minutos), mientras que ciudades de menor productividad como Santiago (USD 2,700) presentan alta congestión (~800 minutos). Sin embargo, México City contradice este patrón: con el segundo PIB per cápita más alto (USD 21,100), registra la mayor congestión de la muestra (~2,833 minutos). Esto sugiere que factores adicionales —densidad poblacional, infraestructura vial, sistemas de transporte público y planificación urbana— modulan significativamente la relación entre economía y movilidad urbana, siendo el PIB per cápita solo una variable entre muchas que explican la congestión.
